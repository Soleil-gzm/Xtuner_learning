# Transformer前馈网络代码实现

## FFN 的结构有多简单？

很简单：**两层线性变换，中间夹一个激活函数。**

公式写出来长这样：

$$
\text{FFN}(x) = \text{Activation}(x \cdot W_1 + b_1) \cdot W_2 + b_2
$$

拆成三步就是：

```
输入 x（d_model 维）
  ↓ 第一层线性变换 W₁（d_model → d_ff，升维）
  ↓ 激活函数（ReLU / GELU）
  ↓ 第二层线性变换 W₂（d_ff → d_model，降维）
输出（d_model 维）
```

注意关键细节：**中间会先升维，再降回原来的维度**。
